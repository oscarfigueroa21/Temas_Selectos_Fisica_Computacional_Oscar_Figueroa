# Ambiente de Trabajo para Astrofísica 


Se usara la misma metodología implementada para la creación del ambiente de trabajo del curso general, usando ahora la **Astrofísica** como línea de trabajo de intéres.



## 1. Crear el ambiente

Ejecuta en PowerShell:

```bash
conda create -n astros python=3.10
```

Esta instrucción crea un nuevo ambiente Conda llamado `astros` e instala **Python 3.10** dentro de él.

El comando puede interpretarse por partes:

```text
conda create
```

crea un nuevo ambiente;

```text
-n astros
```

asigna al ambiente el nombre `astros`;

```text
python=3.10
```

indica la versión de Python que se instalará.

En conjunto:

```text
Crear un ambiente
        ↓
astros
        ↓
Python 3.10
```

Cuando aparezca:

```text
Proceed ([y]/n)?
```

Conda está solicitando confirmación para continuar con la instalación. Escribe:

```text
y
```

y presiona `Enter`.

Al finalizar, el ambiente estará creado. 




![Ambientes creados ](Evidencias/amb.png)

---



## 2 Activar el ambiente

Ejecuta en PowerShell:

```bash
conda activate astros
```

Esta instrucción **activa el ambiente Conda** que creamos anteriormente.

El comando puede interpretarse por partes:

```text
conda activate
```

indica a Conda que queremos comenzar a utilizar un ambiente;

```text
astros
```

es el nombre del ambiente que deseamos activar.

En conjunto:

```text
Activar Conda
      ↓
astros
      ↓
usar Python y las bibliotecas de este ambiente
```

Después de activarlo, el nombre del ambiente deberá aparecer al inicio del prompt:

```text
(astros)
```

Por ejemplo:

```text
(astros) usuario@equipo:~$
```

Esto indica que los comandos `python`, `pip` y las bibliotecas que instalemos se utilizarán dentro de este ambiente.

## 4 Verificar la ruta de Python

Ejecuta en PowerShell:

```powershell
where.exe python
```

Esta instrucción muestra la ubicación del ejecutable de Python disponible en el entorno actual.

En ambos casos, la ruta debe corresponder al ambiente:

```text
astros
```

Por ejemplo, puede aparecer una ruta semejante a:

```text
.../envs/astros/bin/python
```

o, en Windows:

```text
...\\envs\\astros\\python.exe
```

# Instalar Jupyter

Ejecuta en la terminal, con el ambiente `astros` activado:

```bash
conda install -c conda-forge jupyterlab notebook ipykernel
```

Esta instrucción instala las herramientas necesarias para trabajar con **Jupyter Notebooks** dentro del ambiente del curso.

El comando puede interpretarse por partes:

```text
conda install
```

indica a Conda que queremos instalar nuevos paquetes;

```text
-c conda-forge
```

indica que los paquetes se buscarán en el canal `conda-forge`;

```text
jupyterlab notebook ipykernel
```

instala tres componentes:

- `jupyterlab`: entorno interactivo para trabajar con notebooks;
- `notebook`: soporte para archivos `.ipynb`;
- `ipykernel`: permite utilizar nuestro ambiente de Python como kernel de Jupyter.

En conjunto:

```text
Conda
  ↓
instalar Jupyter
  ↓
JupyterLab + Notebook + IPython Kernel
```

Cuando Conda solicite confirmación:

```text
Proceed ([y]/n)?
```

```text
y
```

## Registrar el ambiente como kernel de Jupyter

Después de instalar Jupyter, ejecuta el comando en una sola línea:

```powershell
python -m ipykernel install --user --name astros --display-name "Python 3.10 (Astrofísica)"
```

Esta instrucción registra el ambiente `astros` para que pueda seleccionarse como **kernel** dentro de Jupyter y Visual Studio Code.

Las partes principales son:

```text
python -m ipykernel install
```

ejecuta el módulo `ipykernel` y registra un nuevo kernel;

```text
--user
```

realiza el registro para el usuario actual;

```text
--name astros
```

establece el nombre interno del kernel;

```text
--display-name "Python 3.10 (Astrofísica)"
```

define el nombre que veremos en la interfaz de Jupyter o VS Code.

En conjunto:

```text
ambiente astros
          ↓
registrar como kernel
          ↓
Python 3.10 (Astrofísica)
          ↓
seleccionable en Jupyter y VS Code
```
![Kernel](Evidencias/instal.png)



> **Nota:** instalar Jupyter proporciona las herramientas para trabajar con notebooks; registrar `ipykernel` permite que esos notebooks utilicen específicamente el Python del ambiente `astros`.

#  Instalar herramientas para Astrofísica

Ejecuta en la terminal, con el ambiente `astros` activado:

```bash
conda install -c conda-forge numpy scipy matplotlib pandas astropy astroquery photutils sunpy galpy gala rebound
```
> **¿Por qué este conjunto ?** 

| Biblioteca | Uso |
|---|---|
| Astropy | Proporciona las herramientas fundamentales, incluyendo el manejo de archivos FITS, conversión de coordenadas celestes, unidades físicas, constantes astronómicas y cálculos de cosmología |
| Astroquery | Conectar scripts de Python directamente con bases de datos y archivos astronómicos en línea.|
|Photutils  | Biblioteca especializada en fotometría. |
| Sunpy|Ofrece herramientas especializadas para descargar, analizar y visualizar datos de misiones solares e instrumentos observacionales del Sol |
| Galpy y Gala | Permiten modelar potenciales gravitacionales, calcular órbitas de estrellas, y analizar la estructura y evolución dinámica de galaxias y cúmulos estelares. |
| Rebound | Simulaciones de dinámica de N-cuerpos|
| Pulsarminer y PINT  |Se enfocan en la investigación de púlsares, permitiendo el cronometraje de alta precisión (pulsar timing) y el análisis de señales de radio.|
