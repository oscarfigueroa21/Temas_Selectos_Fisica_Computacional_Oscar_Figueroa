##  Prueba con galaxias

Crea `prueba_astrofísca.py`:

```python
import urllib.request
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.wcs import WCS
from astropy.visualization import (
    ImageNormalize,
    ZScaleInterval,
    PercentileInterval,
    MinMaxInterval,
    LinearStretch,
    SqrtStretch,
    LogStretch,
    AsinhStretch,
)

plt.rcParams["figure.facecolor"] = "white"

DATA_DIR = Path("m51_data")
DATA_DIR.mkdir(exist_ok=True)

BASE_URL = "http://archive.stsci.edu/pub/hlsp/m51/version1"

def descarga_m51(filtro, resolucion="s20", tipo="sci"):
    # Descarga un FITS del mosaico de M51 (b, v, i, h) si no lo tenemos ya.
    nombre = f"h_m51_{filtro}_{resolucion}_drz_{tipo}.fits"
    destino = DATA_DIR / nombre
    if not destino.exists():
        url = f"{BASE_URL}/{nombre}"
        print(f"Descargando {url} ...")
        urllib.request.urlretrieve(url, destino)
        print("Listo:", destino)
    else:
        print("Ya existe:", destino)
    return destino

archivo_v = descarga_m51("v")
archivo_v

with fits.open(archivo_v) as hdul:
    hdul.info()          # resumen de todas las extensiones del archivo
    data = hdul[0].data.astype(float)   # los pixeles, como arreglo de numpy
    header = hdul[0].header             # los metadatos (lo vemos en la sección 4)

print("\nForma de la imagen (filas, columnas):", data.shape)
print("Tipo de dato:", data.dtype)
print("Valor mínimo / máximo:", np.nanmin(data), np.nanmax(data))


stretches = {
    "Lineal": LinearStretch(),
    "Raíz cuadrada (sqrt)": SqrtStretch(),
    "Logarítmica (log)": LogStretch(),
    "Asinh (la más usada en astronomía)": AsinhStretch(),
}

fig, axes = plt.subplots(2, 2, figsize=(10, 10))
for ax, (nombre, stretch) in zip(axes.flat, stretches.items()):
    norm = ImageNormalize(data, interval=ZScaleInterval(), stretch=stretch)
    ax.imshow(data, origin="lower", cmap="gray", norm=norm)
    ax.set_title(nombre)
    ax.axis("off")
plt.suptitle("M51 — mismo intervalo (ZScale), distinto stretch")
plt.tight_layout()
plt.show()

cmaps = ["gray", "viridis", "inferno", "cividis", "hot"]
norm = ImageNormalize(data, interval=ZScaleInterval(), stretch=AsinhStretch())

fig, axes = plt.subplots(1, len(cmaps), figsize=(4 * len(cmaps), 4))
for ax, cmap in zip(axes, cmaps):
    ax.imshow(data, origin="lower", cmap=cmap, norm=norm)
    ax.set_title(cmap)
    ax.axis("off")
plt.suptitle("M51 — mismo stretch (Asinh + ZScale), distinta paleta")
plt.tight_layout()
plt.show()

```
Con el siguiente código vamos a realizar: 
 1. Descargar y leer una imagen FITS real (M51, banda V, *Hubble Space Telescope*).
 2. Visualizarla de forma directa y con distintas escalas/paletas de color.

Para ello, vamos a usar el mosaico de M51 (Whirlpool) tomado con el Advanced Camera for Surveys del HST, disponible públicamente en el archivo de STScI:
